```python
##B. Kueri Bertingkat
# tampilkan hasil tiga kueri berikut

#semua menu kopi dengan tingkat_kepahitan diatas 3       
print("Kopi yang tingkat kepahitannya lebih dari 3: ")
for p in koleksi_menu.find({"tingkat_kepahitan":{"$gt":3}}):
    print(" - ", p["nama_menu"], p["tingkat_kepahitan"])
print()

#semua menu dengan harga diantara harga 15000 dan 30000
print("Menu berharga diantara 15000 sampai 30000")   
for p in koleksi_menu.find({"harga": {"$gt":15000} and {"$lt":30000}}):
    print(" - ",p["nama_menu"],"Rp",p["harga"])
print()

#Menu yang tidak memiliki field pedas level sama sekali
print("menu yang tidak memiliki field tingkat kepedasan")
for p in koleksi_menu.find({"tingkat_kepedasan":{"$exists":False}}):
    print(" - ", p["nama_menu"])
```

```python
##C Pembaruan Data
#Menambahkan field status bernilai "Tersedia" ke semua dokumen dengan update_many

koleksi_menu.update_many(
    {},
    {"$set":{"Status":"Tersedia"}}
)
for p in koleksi_menu.find({}):
    print(p)
    
#Untuk menghapus field yang duplikat misal atau yang tidak dipakai, pakai syntax ini
"""
koleksi_menu.update_many(
    {},
    {"$unset":{"status":""}}
)
"""
```

```python
#Mengubah status menjadi "Habis" khusus 2 menu dengan update_one
koleksi_menu.update_one(
    {"nama_menu":"Rice Bowl Cumi Hitam"},
    {"$set":{"Status":"Habis", "stok":0}}
)
koleksi_menu.update_one(
    {"nama_menu":"Mocha Latte"},
    {"$set":{"Status":"Habis","stok":0}}
)

for p in koleksi_menu.find({}):
    print(" - ", p["nama_menu"],p["Status"],p["stok"])
    
```

```python
##D. Integrasi ke Spark
#Membuat Spark Session
from pyspark.sql import SparkSession
import pandas as pd

spark= SparkSession.builder.appName("Tugas9-MongoDBSpark").master("local[*]").getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

#Mengambil semua dokumen dari MongoDB sebagai list of dict
doc_mongo=list(koleksi_menu.find({}))

#Mengubah menjadi pandas DataFrame
df_pandas=pd.DataFrame(doc_mongo)

#Menyesuaikan tipe data khusus MongoDB (_id, dan tags) agar spark membaca datanya dengan benar
df_pandas["_id"]= df_pandas["_id"].astype(str)
df_pandas["tags"]= df_pandas["tags"].apply(lambda x: ", ".join(x) if isinstance(x, list) else "")

#Konversi ke spark dataframe
df_spark_menu=spark.createDataFrame(df_pandas)

df_spark_menu.printSchema()
df_spark_menu.show()
```

```python
#Mengisi baris data yang kosong
df_spark_bersih = df_spark_menu.fillna({"tingkat_kepahitan":0.0, "tingkat_kepedasan":0.0})
df_spark_bersih.show()
```

```python
from pyspark.sql.functions import avg, count, round
df_spark_menu.groupBy("kategori").agg(
    round(avg("harga"),2).alias("rata-rata harga")
).show()
spark.stop()
print("Sparksession ditutup")
    
```
